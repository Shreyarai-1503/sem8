```python
import numpy as np
from tensorflow import keras
```

```python
# 1) Load data
(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=10000)
```

```python
# 2) Vectorize sequences (multi-hot)
def vec(seqs, dim=10000):
    x = np.zeros((len(seqs), dim), dtype="float32")
    for i, s in enumerate(seqs):
        x[i, s] = 1.0
    return x

x_train, x_test = vec(x_train), vec(x_test)
```

```python
# 3) Build simple DNN
model = keras.Sequential([
    keras.layers.Dense(16, activation="relu", input_shape=(10000,)),
    keras.layers.Dense(16, activation="relu"),
    keras.layers.Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
```

```python
# 4) Train and evaluate
model.fit(x_train, y_train, epochs=4, batch_size=512, validation_split=0.2, verbose=1)
loss, acc = model.evaluate(x_test, y_test, verbose=0)
print(f"Test accuracy: {acc:.4f}")
```

```python

# 5) Sample predictions
p = model.predict(x_test[:5], verbose=0).flatten()
print("Predicted labels:", (p >= 0.5).astype(int))
print("True labels:     ", y_test[:5])
```
